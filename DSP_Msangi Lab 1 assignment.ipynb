```python

import numpy as np
import matplotlib.pyplot as plt

n = np.arange(-5, 11)

delta = (n == 0).astype(float)
step = (n >= 0).astype(float)
ramp = n * step
expo = 0.8**n * step

fig, axes = plt.subplots(1, 4, figsize=(13, 2.6))

for ax, x, title in zip(
    axes,
    [delta, step, ramp, expo],
    ["Impulse", "Step", "Ramp", "Exponential"]
):
    ax.stem(n, x)
    ax.set_title(title)
    ax.set_xlabel("n")
    ax.grid()

plt.tight_layout()
plt.show()
```

```python

import numpy as np
import matplotlib.pyplot as plt

n = np.arange(0, 21)

a1 = 0.5
a2 = -0.8
a3 = 1.1

x1 = a1**n
x2 = a2**n
x3 = a3**n

plt.figure(figsize=(12, 7))

plt.subplot(3, 1, 1)
plt.stem(n, x1)
plt.title("Exponential Signal a = 0.5")
plt.grid()

plt.subplot(3, 1, 2)
plt.stem(n, x2)
plt.title("Exponential Signal a = -0.8")
plt.grid()

plt.subplot(3, 1, 3)
plt.stem(n, x3)
plt.title("Exponential Signal a = 1.1")
plt.grid()

plt.tight_layout()
plt.show()
```

```python

import os
import numpy as np
import matplotlib.pyplot as plt
from scipy.io import wavfile

fname = "my_voice.wav"

if os.path.exists(fname):
    fs, x = wavfile.read(fname)

    if x.ndim > 1:
        x = x.mean(axis=1)

    x = x / np.max(np.abs(x))

else:
    fs = 16000
    t = np.arange(0, 2, 1/fs)
    x = 0.5 * np.sin(2*np.pi*440*t) * np.exp(-t)

t = np.arange(len(x)) / fs

print("Sampling rate:", fs, "Hz")
print("Duration:", len(x)/fs, "seconds")
print("Number of samples:", len(x))

plt.figure(figsize=(12, 4))
plt.plot(t, x)
plt.xlabel("Time (s)")
plt.ylabel("Amplitude")
plt.title("My Voice Recording")
plt.grid()
plt.show()
```
